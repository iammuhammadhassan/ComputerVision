```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("4.png")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
mean = cv2.blur(img, (5, 5))
```

```python
gaussian = cv2.GaussianBlur(img, (5, 5), 0)
```

```python
median = cv2.medianBlur(img, 5)
```

```python
mean_rgb = cv2.cvtColor(mean, cv2.COLOR_BGR2RGB)
gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)
median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(mean_rgb)
plt.title("Mean Filter (5×5)")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter (5×5)")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(median_rgb)
plt.title("Median Filter (5×5)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite("mean_filtered.jpg", mean)
cv2.imwrite("gaussian_filtered.jpg", gaussian)
cv2.imwrite("median_filtered.jpg", median)
```
