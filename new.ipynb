```python
import cv2
import numpy as np
```

```python
img = cv2.imread("mr.jpg")
```

```python
mean = 0
sigma = 30 # You can increase this number to make the noise heavier
gaussian_noise = np.random.normal(mean, sigma, img.shape).astype(np.float32)

# Add noise and clip values to stay within valid color range (0-255)
noisy_gaussian = cv2.add(img.astype(np.float32), gaussian_noise)
noisy_gaussian = np.clip(noisy_gaussian, 0, 255).astype(np.uint8)
```

```python
noisy_sp = img.copy()
prob = 0.05 # 5% of the pixels will be noise

# Generate a random matrix to pick which pixels get corrupted
rand_matrix = np.random.rand(*img.shape[:2])

# Apply salt (white dots)
noisy_sp[rand_matrix < prob / 2] = [255, 255, 255]
# Apply pepper (black dots)
noisy_sp[rand_matrix > 1 - (prob / 2)] = [0, 0, 0]
```

```python
cv2.imwrite("mrbean_gaussian_noise.jpg", noisy_gaussian)
cv2.imwrite("mrbean_salt_pepper_noise.jpg", noisy_sp)

print("Noisy images generated successfully!")
```
